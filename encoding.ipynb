```python
import pandas as pd
#loading the dataframe
df = pd.read_csv("/content/StudentsPerformance.csv")

#checking the column names
print(df.columns)

#label encoding
#it converts categories (text) into numbers
from sklearn.preprocessing import LabelEncoder
      #importing label encoder class

le = LabelEncoder()   #creates an encoder object

#label Encode all categories
categorical_columns = ['gender', 'race/ethnicity', 'parental level of education', 'lunch', 'test preparation course']
for col in categorical_columns:
    df[col] = le.fit_transform(df[col])
print(df.head())

print("\nThus Label Encoding is done!\n\n")

#one-Hot Encoding
df_encoded = pd.get_dummies(df, columns=categorical_columns)
print(df_encoded.head())

print("\nThus One-Hot Encoding is done!\n\n")
```
