```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = sns.load_dataset('tips')
#data exploration
print(df.info)
missing_values = df.isnull().sum()
print(missing_values)
day_counts = df['day'].value_counts()
print(day_counts)
average_tip = df['tip'].mean()
print(average_tip)
maximum_bill = df['total_bill'].max()
print("The maximum total bill is " + str(maximum_bill))
minimum_bill = df['total_bill'].min()
print("The minimum total bill is " + str(minimum_bill))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = sns.load_dataset('tips')
#data filtering
filtered_data = df[df['total_bill'] > 30] #records where total bills is greater than 30
print(filtered_data)
#show only records where time is dinner
dinner_data = df[df['time'] == 'dinner']
print(dinner_data)
#find the average tip given by smokers vs non-smokers
smoker_tip_avg = df.groupby('smoker')['tip'].mean()
print(smoker_tip_avg)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = sns.load_dataset('tips')
#data visualization
#Create a scatter plot showing the relationship between total_bill and tip.
plt.scatter(df['total_bill'],df['tip'])
plt.show()
#creating an histogram of column 'total_bill'
plt.hist(df['total_bill'])
plt.show()
#creating a bar graph showing average tip of each day
average_tip_by_day = df.groupby('day')['tip'].mean()
plt.bar(average_tip_by_day.index,average_tip_by_day.values)
plt.show()
#Create a box plot showing tip distribution for smokers and non-smokers.
smoker_tip_boxplot = df[df['smoker'] == 'Yes']['tip']
non_smoker_tip_boxplot = df[df['smoker'] == 'No']['tip']
plt.boxplot([smoker_tip_boxplot,non_smoker_tip_boxplot])
plt.show()
#Create a correlation heatmap for numeric columns
numeric_columns = df.select_dtypes(include=[np.number])
correlation_matrix = numeric_columns.corr()
sns.heatmap(correlation_matrix,annot=True,cmap='coolwarm')
plt.show()

```

```python
#insights
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = sns.load_dataset('tips')
#finding which day has the highest tip
highest_tip_day = df.groupby('day')['tip'].sum().idxmax()
print("The day with the highest tip is " + str(highest_tip_day))

```
