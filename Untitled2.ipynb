```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
x = [1,36,56,62,55,90]
y = [2,38,55,64,77,88]
y2 = [3,45,49,56,57,78]
plt.plot(x,y,color="red",marker="*")
plt.plot(x,y2,color="blue",marker="*")
fig,a = plt.subplots(2,2)
a[0,0].plot([1,2.6,6], [3.8,5,2])
a[0,0].set_title("Line Plot")
a[0,1].plot([34,56,55],[34,56,66])
a[0,1].set_title("Scatter plot")
a[1,0].plot([1,5,4,2,9],[3,9,5,3,8])
a[1,1].plot([0.1,0.5,0.9,0.2],[0.7,0.4,0.5,0.8])
plt.show()
```
